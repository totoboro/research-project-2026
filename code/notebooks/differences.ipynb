```python
from pathlib import Path

import csv, json, time
from pathlib import Path
from parity_bdd_lab import read_pg, fpjm, chain_multistrategy, zielonka
p=Path('code/data/benchmarks/syntcomp2020-pgame/full_arbiter_unreal3.tlsf.ehoa.oink')
# p=Path('code/data/benchmarks/syntcomp2020-pgame/lilydemo01.tlsf.ehoa.oink')
game=read_pg(p)
solvers=[('FPJM',fpjm),('Chain-MS',chain_multistrategy),('Zielonka',zielonka)]
results={}; times={}
for name,solver in solvers:
    start=time.perf_counter(); results[name]=solver(game); times[name]=time.perf_counter()-start
assert all(r.even_wins==results['FPJM'].even_wins and r.odd_wins==results['FPJM'].odd_wins for r in results.values())
method_edges={n:{0:r.even_strategy,1:r.odd_strategy} for n,r in results.items()}
rows=[]; player_summary={}
for player in (0,1):
    union=set().union(*(method_edges[n][player] for n,_ in solvers))
    common=set.intersection(*(set(method_edges[n][player]) for n,_ in solvers))
    player_summary[str(player)]={
        'FPJM_edges_len':len(method_edges['FPJM'][player]), 
        # 'FPJM_edges':method_edges['FPJM'][player],
        'Chain-MS_edges_len':len(method_edges['Chain-MS'][player]),
        # 'Chain-MS_edges':method_edges['Chain-MS'][player],
        'Zielonka_edges_len':len(method_edges['Zielonka'][player]),
        # 'Zielonka_edges':method_edges['Zielonka'][player],
        # 'CMS_ZLK_union': method_edges['Chain-MS'][player].union(method_edges['Zielonka'][player]),
        'CMS_ZLK_union_len': len(method_edges['Chain-MS'][player].union(method_edges['Zielonka'][player])),
        'CMS_ZLK_union_same_as_FP_len': len(method_edges['Chain-MS'][player].union(method_edges['Zielonka'][player]).difference(method_edges['FPJM'][player])),
        'CMS_ZLK_union_same_as_FP': method_edges['Chain-MS'][player].union(method_edges['Zielonka'][player]).difference(method_edges['FPJM'][player]),
        'common_to_all_three':len(common)
        }
    for source,target in sorted(union):
        present={n:(source,target) in method_edges[n][player] for n,_ in solvers}
        if len(set(present.values()))==1: continue
        rows.append({'player':player,'source':source,'source_owner':game.owner[source],'priority':game.priority[source],'target':target,**{n:present[n] for n,_ in solvers}})

print(rows)
print(player_summary)
```
